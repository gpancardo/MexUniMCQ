## Reducing redundant topics

```python
import json
import re

def reduce_topic(t):
    if not t:
        return t

    t = t.strip()

    # Preguntas introductorias
    t = re.sub(r"^¿Qué es (.+?)\?$", r"\1", t)

    # Slash: subtema didáctico
    if "/" in t:
        t = t.split("/", 1)[0].strip()

    # Dos puntos explicativos
    if ":" in t:
        t = t.split(":", 1)[0].strip()

    # Guion con subtema (materias claras)
    if " - " in t:
        t = t.split(" - ", 1)[0].strip()

    # Normalizaciones curriculares aceptadas
    if t == "Velocidad de reacción y factores que influyen en ella":
        t = "Cinética química"

    if t == "Ética de la ciencia y la tecnología":
        t = "Ética"

    return t

records = []
with open("MexUniMCQ.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        obj["topic_reduced"] = reduce_topic(obj.get("topic"))
        records.append(obj)

with open("MexUniMCQ_reduced_topics.jsonl", "w", encoding="utf-8") as f:
    for obj in records:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("Listo. Todas las reducciones aplicadas.")

```

```python
import re

def reduce_topic(t):
    if not t:
        return t

    t = t.strip()

    # ¿Qué es X?
    t = re.sub(r"^¿Qué es (.+?)\?$", r"\1", t)

    # Eliminar TODO lo que esté después de / (con o sin espacios)
    t = re.sub(r"\s*/.*$", "", t)

    # Eliminar TODO lo que esté después de :
    t = re.sub(r"\s*:.*$", "", t)

    # Eliminar subtema tras guion
    t = re.sub(r"\s-\s.*$", "", t)

    # Normalizaciones curriculares aceptadas
    if t == "Velocidad de reacción y factores que influyen en ella":
        t = "Cinética química"

    if t == "Ética de la ciencia y la tecnología":
        t = "Ética"

    return t.strip()

records = []
with open("MexUniMCQ.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        obj["tema"] = reduce_topic(obj.get("tema"))
        records.append(obj)

with open("MexUniMCQ_reduced_topics.jsonl", "w", encoding="utf-8") as f:
    for obj in records:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("Listo. Reducción correcta aplicada.")
```

## ID Generation

```python
# List temas and materias from MexUniMCQ_reduced_topics.jsonl and assign three digit IDs to each one. It will be used to create a 11 digit ID per question
temas = set()
materias = set()
with open('MexUniMCQ_reduced_topics.jsonl', 'r', encoding='utf-8') as f:
        for line in f:
            data = json.loads(line)
            temas.add(data['tema'])
            materias.add(data['materia'])
temas = sorted(list(temas))
materias = sorted(list(materias))
tema_id_map = {tema: f"{i:03d}" for i, tema in enumerate(temas, start=1)}
materia_id_map = {materia: f"{i:03d}" for i, materia in enumerate(materias, start=1)}
```

```python
records = []

correcta_map = {"a": 1, "b": 2, "c": 3, "d": 4}

with open("MexUniMCQ_reduced_topics.jsonl", "r", encoding="utf-8") as f:
    for idx, line in enumerate(f, start=1):
        obj = json.loads(line)

        question_number = f"{idx:04d}"
        materia_id = materia_id_map.get(obj["materia"], "000")
        tema_id = tema_id_map.get(obj["tema"], "000")

        correcta = obj.get("correcta", "").lower()
        alt_idx = correcta_map.get(correcta)

        if alt_idx is None:
            raise ValueError(f"Valor inválido en 'correcta': {obj.get('correcta')}")

        unique_id = f"{question_number}{materia_id}{tema_id}{alt_idx}"

        obj["unique_id"] = unique_id

        records.append(obj)

with open("MexUniMCQ_with_ids.jsonl", "w", encoding="utf-8") as f:
    for obj in records:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")
```
